```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
   "Department": ["HR", "Finance", "IT", "Sales"],
   "Absence": [5, 8, 12, 7]
})
df = df.sort_values("Absence")
plt.bar(df["Department"], df["Absence"],
       color=["Grey"])
plt.title("Employee Absence")
plt.xlabel("Department")
plt.ylabel("Absence %")
plt.grid(False)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("IT Has the Highest Absence — 12%")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.annotate("Highest",
            xy=("IT", 12),
            xytext=("Finance", 13),
            arrowprops=dict(arrowstyle="->"))
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.show()
```

```python
arabic = ["الموارد البشرية", "المالية", "تقنية المعلومات", "المبيعات"]

plt.figure(figsize=(9,5))
plt.bar(arabic, df["Absence"], color="steelblue")
plt.title("تقنية المعلومات لديها أعلى نسبة غياب — 12%")
plt.xlabel("القسم")
plt.ylabel("نسبة الغياب (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()
```
