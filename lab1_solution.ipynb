```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
   "Region": ["Riyadh", "Makkah", "Madinah", "Eastern", "Asir"],
   "Adoption": [85, 72, 91, 65, 78]
}

df = pd.DataFrame(data)

df
df_sorted = df.sort_values("Adoption")

plt.figure(figsize=(8,5))
plt.bar(df_sorted["Region"], df_sorted["Adoption"])
plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.title("Digital Adoption by Region — Sorted")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Region"], df["Adoption"])
plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.title("Digital Adoption by Region — Unsorted")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,2))

sns.heatmap(
   [df["Adoption"]],
   annot=True,
   xticklabels=df["Region"],
   yticklabels=["Adoption"],
   cmap="Blues"
)

plt.title("Digital Adoption by Region — Heatmap")
plt.show()
```

```python
plt.figure(figsize=(7,7))

plt.pie(
   df["Adoption"],
   labels=df["Region"],
   autopct="%1.1f%%"
)

plt.title("Digital Adoption by Region — Pie")
plt.show()

```

```python
plt.figure(figsize=(8,5))

plt.scatter(
   df["Region"],
   df["Adoption"],
   s=df["Adoption"] * 20
)

plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.title("Digital Adoption by Region — Bubble")
plt.xticks(rotation=45)
plt.show()
```
