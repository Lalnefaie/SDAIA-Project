```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

q1 = pd.DataFrame({
   "Region": ["Najran", "Eastern", "Jazan", "Asir", "Makkah", "Madinah"],
   "Adoption": [58, 60, 64, 67, 72, 78]
})

q1 = q1.sort_values("Adoption")

plt.figure(figsize=(8,5))
plt.bar(q1["Region"], q1["Adoption"])
plt.axhline(q1["Adoption"].mean(), linestyle="--")
plt.xlabel("Region")
plt.ylabel("Digital Adoption (%)")
plt.title("Regional Adoption — Lowest to Highest")
plt.bar_label(plt.gca().containers[0])
plt.show()
```

```python
q2 = pd.DataFrame({
   "Month": ["January", "February", "March", "April", "May", "June"],
   "Adoption": [58, 61, 63, 66, 69, 72]
})

plt.figure(figsize=(8,5))
plt.plot(q2["Month"], q2["Adoption"], marker="o")
plt.axhline(65, linestyle="--")
plt.ylabel("Digital Adoption (%)")
plt.title("National Digital Adoption Trend")
plt.show()
```

```python

fig = px.line(
   q2,
   x="Month",
   y="Adoption",
   markers=True,
   title="National Digital Adoption Trend"
)

fig.add_hline(y=65, line_dash="dash")

fig.show()
```

```python
q3 = pd.DataFrame({
   "Category": ["Digital Services"]*8 + ["Support"]*8 + ["Payments"]*8,
   "CompletionTime": [
       18,20,21,22,23,25,26,40,
       12,14,15,16,17,18,19,30,
       20,21,22,24,25,26,28,35
   ]
})

sns.boxplot(data=q3, x="Category", y="CompletionTime")

plt.title("Completion Time by Category")
plt.ylabel("Completion Time")
plt.show()
```
